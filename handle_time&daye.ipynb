```python
import numpy as np
import pandas as pd
```

```python
date  = pd.read_csv('lunar.csv')
```

```python
date.head()
```

```python
date.info()
```

```python
date['Calendar Date'] = pd.to_datetime(date['Calendar Date'], errors='coerce')
```

```python
date.info()
```

<h4>Extract Year</h4>

```python
date['date_year'] = date['Calendar Date'].dt.year
date.head()
```

<h4>Extract month</h4>
<bg.color='Red'>

```python
date['date_month'] = date['Calendar Date'].dt.month
date.head()
```

```python
date['date_day'] = date['Calendar Date'].dt.day
date.head()
```

```python
date['date_day_name'] = date['Calendar Date'].dt.day_name()
date.head()
```

```python
date['date_week'] = date['Calendar Date'].dt.isocalendar().week
date.head()
```
