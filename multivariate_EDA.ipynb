```python
import pandas as pd
import seaborn as sns
```

```python
titanic = pd.read_csv('train (4).csv')
```

```python
Iris = pd.read_csv('IRIS.csv')
```

```python
tips = pd.read_csv('tips.csv')
```

<h2>1. Scatterplot(numercial-numerical)</h2>

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'])
```

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

<h2>2. Barplot(Numerical-Categorical)</h2>

```python
titanic.head()
```

```python
sns.barplot(x=titanic['Pclass'],y=titanic['Age'])
```

```python
sns.barplot(x=titanic['Pclass'],y=titanic['Age'],hue=titanic['Sex'])
```

<h2>2. Boxplot(Numerical - Categorical)</h2>

```python
sns.boxplot(x=titanic['Sex'],y=titanic['Age'],hue=titanic['Survived'])
```

```python
sns.boxplot(data=titanic,x='Sex',y='Age',hue='Survived')
```

<h2>4. Distplot</h2>

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'],hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'],hist=False)
```

<h2>HeatMap(Categorical - Categorical)</h2>

```python
titanic.head()
```

```python
pd.crosstab(titanic['Pclass'],titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
titanic.groupby('Pclass')['Survived'].mean()
```

<h2>ClusterMap (categorical - Categorical)</h2>

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```
